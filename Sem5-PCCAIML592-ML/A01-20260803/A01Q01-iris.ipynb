Using Pandas, perform DataFrame operations and Statistical analysis on the Iris Dataset.

```python
import pandas as pd
```

We can load the iris dataset using `from sklearn.datasets import load_iris` which is an inbuilt function in scikit-learn that loads the famous Iris Flower dataset. But, here we are loading using `read_csv("iris-data.csv")`

The `iris` object contains:
- 150 flower samples
- 4 features:
    - sepal length
    - sepal width
    - petal length
    - petal width
- 3 classes/species:
    - setosa
    - versicolor
    - virginica

```python
iris_df = pd.read_csv(
    "iris-data.csv", 
    header=None,
    names=["sepal_length", "sepal_width", "petal_length", "petal_width", "species"])
```

```python
print("\nIris Dataset")

print("\nHead\n", iris_df.head(), sep ="")

print("\nTail\n", iris_df.tail(), sep ="")

print("\nShape:", iris_df.shape)

print("\nIndex:", iris_df.index)

print("\nDescribe\n", iris_df.describe())

print("\nMissing Values\n", iris_df.isna().sum(), sep ="")
```

```python
print("\nMean\n", iris_df.mean(numeric_only=True), sep ="")

print("\nMedian\n", iris_df.median(numeric_only=True), sep ="")

print("\nMode\n", iris_df.mode(numeric_only=True), sep ="")

print("\nMaximum\n", iris_df.max(numeric_only=True), sep ="")

print("\nCount\n", iris_df.count(), sep ="")

print("\nStandard Deviation\n", iris_df.std(numeric_only=True), sep ="")

print("\nSum\n", iris_df.sum(numeric_only=True), sep ="")
```
