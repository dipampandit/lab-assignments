Using Pandas, perform DataFrame operations and Statistical analysis on the Student Alcohol Consumption Dataset.

```python
import pandas as pd
```

```python
alcohol_df = pd.read_csv("student-mat.csv")
```

```python
print("\nStudent Alcohol Consumption Dataset")

print("\nHead\n", alcohol_df.head(), sep="")

print("\nTail\n", alcohol_df.tail(), sep="")

print("\nShape:", alcohol_df.shape)

print("Index:", alcohol_df.index)

print("\nDescribe", alcohol_df.describe(), sep="")

print("\nSorted by G3")
print(alcohol_df.sort_values(by="G3"))

print("\nAfter Dropping Address Column")
print(alcohol_df.drop(columns="address").head())

print("\nMissing Values\n")
print(alcohol_df.isna().sum(), sep="")
```

```python
print("\nMean\n", alcohol_df.mean(numeric_only=True), sep="")

print("\nMedian\n", alcohol_df.median(numeric_only=True), sep="")

print("\nMode\n", alcohol_df.mode(), sep="")

print("\nMaximum\n", alcohol_df.max(), sep="")

print("\nCount\n", alcohol_df.count(), sep="")

print("\nStandard Deviation\n", alcohol_df.std(numeric_only=True), sep="")

print("\nSum\n", alcohol_df.sum(numeric_only=True), sep="")
```
